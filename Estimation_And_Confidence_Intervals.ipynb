###Central limit theorem
It is the one of fundamental concept of stats with significant to the datascience. It states that the distribution of the sample mean of any independent random variable will be normal or nearly normal, if the saple size is large enough.
####Estimation
According to the Central Limit Theorem(CLT) the sample mean Xbar is a good point estimator of the population mean(mu). As the sample size increases the sample mean gets closer and closer to the true polpulation mean.

####Confidence intervals:
Confidence interval are the intervel estimates that provide a range of plausible values for a population parameter like mean. The CLT is the theoratical justification for constructing many common confidence intervals, especially for the population mean

```python
# imporing the required libraries
import numpy as np
import scipy.stats

#data for print head durability in million of characters
data = np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
n=len(data)
alpha=0.01# for 99% confidence interval


# Task a: Build 99% confidence intervel using sample Standard Deviation

sample_mean = np.mean(data)#To find the mean of the data
sample_std = np.std(data, ddof=1)#To find standard deviation
degreeOfFreedom = n-1
t_critical = scipy.stats.t.ppf(1-alpha/2, df=degreeOfFreedom)
standErrMeanSample = sample_std/np.sqrt(n)
marginOfErrorSample = t_critical*standErrMeanSample
confidenceIntervalSample = [sample_mean - marginOfErrorSample, sample_mean + marginOfErrorSample]

print(f'Sample Mean: {sample_mean} in miilon chars')
print(f'Sample Standard Deviation: {sample_std} in million chars')
print(f'Degree of Freedom: {degreeOfFreedom}')
print(f'T-critical value: {t_critical}')
print(f'Standard Error of the Mean: {standErrMeanSample} in million chars')
print(f'Margin of Error: {marginOfErrorSample} in million chars')
print(f'99% Confidence Interval: {confidenceIntervalSample} in million chars')

%
```

```python
#Task b: Build 99% confidence interval using known population Standard Deviation

Known_Population_Std = 0.2
z_critical = scipy.stats.norm.ppf(1-alpha/2)
standErrMeanPopulation = Known_Population_Std/np.sqrt(n)
knownMarginOfErrorPopulation = z_critical*standErrMeanPopulation
ConfidenceIntervelKnown= (sample_mean - knownMarginOfErrorPopulation, sample_mean + knownMarginOfErrorPopulation)

print(f'Sample Mean: {sample_mean} in miilon chars')
print(f'Known Population Standard Deviation : {Known_Population_Std}')
print(f'Z-critical value: {z_critical}')
print(f'Standard Error of the Mean: {standErrMeanPopulation} in million chars')
print(f'Margin of Error: {knownMarginOfErrorPopulation} in miilon chars')
print(f'99% Confidence Interval: {ConfidenceIntervelKnown}')
```
